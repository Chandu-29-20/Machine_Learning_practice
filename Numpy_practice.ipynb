<span>Question 3: Element-wise Arithmetic and Aggregations You are given two 1D NumPy arrays representing the unit prices (in ₹) and quantities sold for 5 products:<br>
Write a program to compute and print: <br>
An array representing the total revenue generated by each individual product. <br>
The overall combined revenue across all 5 products. <br>
The average revenue per product. <br>
The index position of the highest-earning product using np.argmax() </span>

```python
prices = np.array([120, 250, 80, 450, 300]) 
quantities = np.array([4, 2, 10, 1, 3])
revenue=prices*quantities
print("Revenue of each product:", revenue)
tt_rev=np.sum(revenue)
print("Overall Revenue:",tt_rev)
avg_rev=np.mean(revenue)
print("Average Revenue:",avg_rev)
idx=np.argmax(revenue)
print("Highest-earning product:",idx)
```

<span>Question 4: Boolean Masking and Data Cleaning. A weather monitoring station recorded daily rainfall (in mm) over 10 days. Due to sensor glitches, invalid negative values were logged in the dataset:<br>
Using boolean indexing: <br>
Extract and print a new array containing only the valid (non-negative) rainfall readings. <br>
Replace all negative values in the rainfall array with 0.0 and print the cleaned array. <br>
Count and print how many days recorded heavy rainfall (strictly greater than 10.0 mm).</span>

```python
rainfall = np.array([12.5, -1.0, 0.0, 18.2, -99.0, 5.4, 22.1, 0.0, -1.0, 14.8])
val_rf=rainfall[rainfall>=0]
print("Valid rainfall readings:",val_rf)
rainfall[rainfall<0]=0.0
print(rainfall)
count_rf=np.sum(rainfall>10.0)
print("Heavy Rainfall:",count_rf)
```

<span>Question 5: Conditional Array Operations with np.where() . Generate a 1D array of 12 random integers between 20 and (remember to set 100 representing student exam scores np.random.seed(10) ). Create and print a new array named "Pass" to scores greater than or equal updated_scores where any student who scored below receives 5 grace marks, while scores of 40 and above remain unchanged.</span>

```python
s=np.random.randint(20,101,size=12)
print("Scores:",s)
status=np.where(s>= 40, "Pass", "Fail")
print("Status:",status)
upd_sc=np.where(s<40,s+5,s)
print("updated_scores:",upd_sc)
```

<span>Question 7: Feature Scaling via Broadcasting You are given a 5 × 3 dataset representing 5 properties with 3 features ([Area_sqft, Bedrooms, Price_Lakh]) <br>
Normalize every feature column independently so that all values fall between Max normalization and NumPy broadcasting: X − Xmin Xnorm = Print the normalized Xmax − Xmin 5 × 3 array rounded to 3 decimal places. </span>

```python
houses = np.array([ [1200, 2, 55], [1800, 3, 85], [950,  1, 40], [2400, 4, 120], [1500, 3, 70] ])
X_min = np.min(houses, axis=0)
X_max = np.max(houses, axis=0)
X_norm = (houses - X_min) / (X_max - X_min)
print(np.round(X_norm, 3))
```

<span>Question 8: Combining and Stacking Arrays 0 and 1 using MinA retail company tracks morning and evening shift sales (in units) across 3 store branches for 2 products:<br>
Write code to: <br>
Combine morning_sales and evening_sales horizontally to form a 3 × 4 matrix representing full-day sales across all 3 branches. <br>
Append a 4th branch's full-day sales record [12, 16, 15, 19] as a new row at the bottom to create a 4 × 4 matrix. <br>
Print the final 4 × 4 matrix and the total units sold in each of the 4 columns.</span>

```python
morning_sales = np.array([ [15, 20], [10, 12], [25, 30] ]) 
evening_sales = np.array([ [18, 22], [14, 15], [20, 28] ])
a=np.hstack((morning_sales,evening_sales))
print(a)
b=[12, 16, 15, 19]
c=np.vstack((a,b))
print(c)
total=np.sum(c,axis=0)
print("Total Units Sold:",total)
```

<span>Question 9: Unique Elements and Frequency Distribution A customer feedback survey collected ratings (on a scale of 1 to 5) from 20 customers:<br>Write a program to: <br>
Find all unique rating values and the exact frequency count of each rating using Calculate the percentage share of each rating out of the total responses. np.unique() . Print a clean summary displaying each rating, its count, and its percentage (for example: 1: 2 responses (10.0%) ).</span>

```python
ratings = np.array([5, 4, 3, 5, 2, 5, 4, 1, 3, 5, 4, 4, 2, 5, 3, 4, 5, 1, 4, 3])
a,fr=np.unique(ratings,return_counts=True)
pe= (fr / len(ratings)) * 100
print("Unique rating values:", a)
print("Frequency count:", fr)
print("Percentage:", pe)
for i in range(len(a)):
    print(f"Rating {a[i]}: {fr[i]} responses ({pe[i]:.1f}%)")
```

<span>Question 10: Indirect Sorting and Top-K Selection Rating You are given two aligned 1D arrays containing employee IDs and their annual performance scores:<br>
Using np.argsort() : Obtain the index order that sorts scores from highest to lowest (descending order). Use those sorted indices to reorder both emp_ids and scores . Print the Top 3 Employees (Employee ID and Score) in descending order of performance.</span>

```python
emp_ids = np.array(["E101", "E102", "E103", "E104", "E105", "E106"]) 
scores  = np.array([78, 92, 65, 88, 95, 71])
t=np.argsort(scores)[::-1]
print("Descending index order:",t)
emp= emp_ids[t]
score= scores[t]
print("Sorted Employee IDs:", emp)
print("Sorted Scores:", score)
print("Top 3 Employees:")
for i in range(3):
    print(f"{emp[i]} : {score[i]}")
```
